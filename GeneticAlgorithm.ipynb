```python
import random

QUEUE_NS, QUEUE_EW = 65, 15
CYCLE_TIME, MIN_GREEN = 90, 15
POP_SIZE, GENERATIONS, MUTATION_RATE = 10, 30, 0.2


def get_fitness(chrom):
    delay = (QUEUE_NS * chrom[1]) + (QUEUE_EW * chrom[0])
    return 1 / (delay + 1)

pop = [
    [g := random.randint(MIN_GREEN, CYCLE_TIME - MIN_GREEN), CYCLE_TIME - g]
    for _ in range(POP_SIZE)
]

for _ in range(GENERATIONS):
    pop.sort(key=get_fitness, reverse=True)
    next_pop = pop[:2]

    while len(next_pop) < POP_SIZE:
        p1, p2 = random.choice(pop[: POP_SIZE // 2]), random.choice(
            pop[: POP_SIZE // 2]
        )
        c_ns = int((p1[0] + p2[0]) / 2)
        child = [c_ns, CYCLE_TIME - c_ns]

        if random.random() < MUTATION_RATE:
            new_ns = child[0] + random.choice([-5, 5])
            if MIN_GREEN <= new_ns <= (CYCLE_TIME - MIN_GREEN):
                child = [new_ns, CYCLE_TIME - new_ns]

        next_pop.append(child)
    pop = next_pop

pop.sort(key=get_fitness, reverse=True)
best = pop[0]
min_delay = (QUEUE_NS * best[1]) + (QUEUE_EW * best[0])

print(f"=== COLAB OPTIMIZATION RUN ===")
print(f"Inputs: NS Queue = {QUEUE_NS} cars | EW Queue = {QUEUE_EW} cars")
print(f"Optimized Plan: [NS Green: {best[0]}s, EW Green: {best[1]}s]")
print(f"Total Combined Delay: {min_delay} vehicle-seconds")
```
